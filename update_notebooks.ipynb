# Automating File Updates

With this script, you can automatically download updated notebooks and replace the existing ones in your local repository. While this can also be done manually, this method is more efficient. All you need to do is adjust the `urls` and `local_paths` to match your needs.


```python
import requests
import os

def download_files(urls, local_paths, replace=True):
    """
    Downloads files from GitHub etc. and optionally replaces existing files.
    
    Parameters:
    - urls (list of str): List of GitHub raw URLs to download the notebooks.
    - local_paths (list of str): List of local file paths where the notebooks should be saved.
    - replace (bool): If True, existing files will be replaced. If False, only download if the file doesn't exist.
    
    Returns:
    - None
    """
    if len(urls) != len(local_paths):
        raise ValueError("The number of URLs and local paths must match.")
    
    for url, local_path in zip(urls, local_paths):
        try:
            # Check if file already exists
            if not replace and os.path.exists(local_path):
                print(f"File already exists and 'replace' is False: {local_path}")
                continue
            
            # Download the notebook from GitHub
            response = requests.get(url)
            response.raise_for_status()  # Raise error if request failed

            # Save the notebook to the specified path
            with open(local_path, "wb") as file:
                file.write(response.content)

            print(f"File successfully downloaded: {local_path}")
        except requests.exceptions.RequestException as e:
            print(f"Error downloading {url}: {e}")
```
