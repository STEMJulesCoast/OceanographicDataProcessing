It contains twelve months, latitude and longitude coordinates in ascending order,
and no missing values. No download, NetCDF file, or Cartopy installation is needed.


```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

```

```python
# We Create a small synthetic dataset
time = np.arange("2025-01", "2026-01", dtype="datetime64[M]")
lat = np.arange(-20, 21, 5)
lon = np.arange(40, 101, 5)
month_index, latitude, longitude = np.meshgrid(
    np.arange(12), lat, lon, indexing="ij"
)

u = 4 + 2 * np.sin(2 * np.pi * month_index / 12) + (longitude - 40) / 60
v = 2 + latitude / 40 + 0.5 * np.cos(2 * np.pi * month_index / 12)

ds = xr.Dataset(
    data_vars={
        "eastward_wind": (("time", "lat", "lon"), u),
        "northward_wind": (("time", "lat", "lon"), v),
    },
    coords={"time": time, "lat": lat, "lon": lon},
    attrs={"description": "Synthetic wind data for a coding exercise"},
)
ds["eastward_wind"].attrs["units"] = "m s-1"
ds["northward_wind"].attrs["units"] = "m s-1"
ds["lat"].attrs["units"] = "degrees_north"
ds["lon"].attrs["units"] = "degrees_east"
```
