# Functions / Modules

Functions are a foundational concept in programming and offer numerous advantages.

**Modularity:** Functions divide a program into smaller parts, making it clearer and easier to maintain. For Ekman calculations, processes such as wind stress calculation can have their own function.  
**Reusability:** A defined function can be used multiple times, eliminating repetitive code. 
**Error Reduction:** Testing individual functions can reduce errors when updating code. If one function works, it doesn't need retesting when you adjust other parts of your code.  
**Parameterization:** Functions allow parameter adjustments easily.   
**Readability:** Descriptive function names make the main code more understandable.    
**Scalability:** Functions can be updated or extended easily.

Function Name: The function name should describe its purpose. For example, `multiply` indicates that the function multiplies two values.

Return Statement: The function returns a value that can be used elsewhere in the code.

```python
# Function to multiply two numbers
def multiply(a,b):
    """Return the product of a and b."""
    result = a*b
    return result
```

### Testing Functions

Testing ensures that functions behave as expected and produce correct results.
Even small tests help detect logical errors early and make code more reliable.
By defining a few assertions (tests that must be true), you can quickly verify your function outputs.
If a test fails, Python raises an AssertionError, which indicates that the function needs correction.

Testing is an essential step toward writing reproducible and trustworthy scientific code.


```python
# --- Test add() ---
assert add(2, 3) == 5
assert add(-1, 1) == 0

# --- Test multiply() ---
assert multiply(2, 3) == 6
assert multiply(-2, 3) == -6

# --- Test divide() ---
assert divide(6, 3) == 2
try:
    divide(1, 0)
    raise AssertionError("Expected ValueError for division by zero.")
except ValueError:
    pass  # Expected behavior

print("All tests passed successfully ✅")
```


### Why Testing Matters – Detecting Logic Errors

Even if code runs without syntax errors, it might still produce *wrong results* due to logic errors.  
Testing helps detect such issues automatically.  
Below, we define a simple function with a hidden logic error and see how testing exposes it.


```python

# Faulty function: should multiply but actually adds
def multiply_wrong(a, b):
    """Incorrect version of multiply: adds instead of multiplies."""
    return a + b  #logic error

# Correct function (for later comparison)
def multiply(a, b):
    """Return the product of a and b."""
    return a * b
```

```python

# Test both functions
assert multiply(2, 3) == 6  # passes
assert multiply_wrong(2, 3) == 6  # should fail
```

The test failed because `multiply_wrong(2, 3)` returned **5 instead of 6**.  
This illustrates that even small logic mistakes can lead to incorrect scientific results —  
and why **automated tests** are essential for reliable data analysis.
